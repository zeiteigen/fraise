# Compare tokenizers

Use saved tokenizer paths explicitly. Loading preserves their saved behavior. Install notebook tools with `uv sync --extra notebooks`.


```python
from IPython.display import HTML, display
from fraise.paths import data_root
from fraise.tokenizer.loading import load_tokenizer
from fraise.preview.tokenization import render_tokenization

paths = sorted((data_root() / "tokenizers").glob("fraise-*.json"))
for number, path in enumerate(paths, 1):
    print(number, path.name)
```

```python
text = input("Text to tokenize: ")
selection = input("Tokenizer numbers, comma-separated [1]: ").strip() or "1"
numbers = list(dict.fromkeys(int(part.strip()) for part in selection.split(",")))
if any(number < 1 or number > len(paths) for number in numbers):
    raise ValueError(f"Select tokenizer numbers from 1 to {len(paths)}")
for number in numbers:
    path = paths[number - 1]
    tokenizer = load_tokenizer(path)
    display(HTML(render_tokenization(text, tokenizer, path.name)))
    encoding = tokenizer.encode(text)
    print("IDs:", encoding.ids)
    print("Tokens:", encoding.tokens)
```
